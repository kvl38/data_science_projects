# Разработка системы персонализированных предложений для клиентов интернет-магазина «В один клик»

**Описание проекта:**

Интернет-магазин “В один клик”, специализирующийся на продаже широкого ассортимента товаров (от детских товаров до продуктов питания), столкнулся с проблемой снижения активности покупателей. Традиционные методы привлечения новых клиентов стали менее эффективными, так как о магазине уже знает значительная часть целевой аудитории. Компания планирует повысить эффективность за счет удержания постоянных клиентов через персонализированные предложения. Для этого необходимо разработать аналитическое решение на основе данных о покупательском поведении.

**Цель проекта:**

Создать модель прогнозирования снижения покупательской активности клиентов и разработать систему персонализированных предложений для поддержания и увеличения их активности в течение следующих трех месяцев.

**Задачи:**

- Маркировка уровней финансовой активности клиентов
- Сбор данных по 4 группам признаков:
     * Коммуникация с клиентом
     * Продуктовое поведение
     * Финансовое поведение
     * Поведение на сайте
- Создание модели прогнозирования снижения покупательской активности на 3 месяца
- Выделение сегментов покупателей на основе анализа данных
- Разработка персонализированных предложений для каждого сегмента
- Подготовка документации по результатам проекта

```python
# Функция для вывода графика количественных признаков
def graph_quant_feature(df, col_name, col_title):
    
    """
    Функция в качестве аргументов принимает датафрейм, название столбца, для которого необходимо вывести график, 
    и общее название для 2-х графиков. Она выводит гистограмму и диаграмму размаха для полученных данных
    """
    
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12,6))
    
    sns.histplot(df[col_name], ax=axes[0], palette='deep', edgecolor='black')
    axes[0].set_xlabel(f'Количественный признак для столбца \'{col_name}\'')
    axes[0].set_ylabel('Количество клиентов')
    axes[0].set_title('Гистограмма')
    axes[0].grid(True)
    
    sns.boxplot(data=df[col_name], orient='h', ax=axes[1], color='orange')
    axes[1].set_xlabel(f'Количественная характеристика для столбца \'{col_name}\'')
    axes[1].set_title('Диаграмма размаха')
    
    plt.suptitle(col_title)
    plt.show()
```

```python
# Функция для вывода графика категориальных признаков
def graph_cat_feature(df, col_name, col_title):
      
    """
    Функция в качестве аргументов принимает датафрейм, название столбца, для которого необходимо вывести график, 
    и название самого графика. Она выводит столбчатый график для полученных данных
    """
      
    plt.figure(figsize=(6, 6))
    sns.countplot(data=df, y=col_name, palette='deep', edgecolor='black')
    plt.title(col_title)
    plt.xlabel('Количество клиентов')
    plt.ylabel(f'Значение столбца \'{col_name}\'')
    plt.xticks(rotation=0)
    plt.show()
```

```python
# Функция построения диаграммы зависимости вероятности снижения активности от выручки по каждому признаку
def scatterplots(df, columns):
    
    """
    Функция принимает датафрейм и список названий столбцов, по которым необходимо построить диаграммы рассеяния, 
    отражающие зависимость вероятности снижения активности от выручки
    """
    
    for col_name in columns:
        plt.figure(figsize=(10, 8))
        sns.scatterplot(data=df, y='прибыль', x='вероятность_класса_снизилась', hue=col_name)
        plt.title(f'Зависимость вероятности снижения активности от выручки')
        plt.xlabel('Вероятность снижения активности')
        plt.ylabel('Прибыль')
        plt.show()
```

## Предобработка данных

### Переименование столбцов

```python
# Преобразуем названия столбцов к нижнему регистру и формату snake_case
dfs = [market_file_df, market_money_df, market_time_df, money_df]

for df in dfs:
    df.columns = df.columns.str.replace(' ', '_').str.lower()
    
    # Проверяем корректность переименования столбцов
    print(df.columns)
```

### Преобразование значений датафрейма к нижнему регистру

```python
# Приводим значения к нижнему регистру
market_file_df['покупательская_активность'] = market_file_df['покупательская_активность'].str.lower()
market_file_df['популярная_категория'] = market_file_df['популярная_категория'].str.lower()

# Приведем значения столбца «популярная_категория» и «популярная_категория» к стилю snake_case
market_file_df['покупательская_активность'] = market_file_df['покупательская_активность'].str.replace(' ', '_')
market_file_df['популярная_категория'] = market_file_df['популярная_категория'].str.replace(' ', '_')

# Проверяем корректность преобразований
market_file_df[['покупательская_активность', 'популярная_категория']].head(3)
```

### Устранение грамматических ошибок

```python
# Исправляем найденную грамматическую ошибку в market_time_df
market_time_df['период'].replace('предыдцщий_месяц', 'предыдущий_месяц', inplace=True)

# Проверяем корректность преобразований
market_time_df['период'].unique()
```

```python
# Исправляем найденную грамматическую ошибку в market_file_df
market_file_df['тип_сервиса'].replace('стандартт', 'стандарт', inplace=True)

# Проверяем корректность преобразований
market_file_df['тип_сервиса'].unique()
```

```python
# Изменим название столбца в market_time_df
market_time_df = market_time_df.rename(columns={'минут': 'минуты'})

# Проверяем корректность преобразований
market_time_df.head(2)
```

Количество значений класса *прежний_уровень* почти в два раза превышает количество значений класса *снизилась*. Примем этот дисбаланс во внимание при разбиении на выборки.

```python
# Разделяем исходные данные на обучающую и тестовую выборки
X = market_file_df_merged.drop(['покупательская_активность','id'], axis=1)
y = market_file_df_merged['покупательская_активность']

X_train, X_test, y_train, y_test = train_test_split(X, 
                                                    y , 
                                                    test_size=TEST_SIZE, 
                                                    random_state=RANDOM_STATE,
                                                    stratify=y)

# Рассмотрим размеры сформированных выборок
X_train.shape, X_test.shape
```

```python
# Закодируем целевой признак
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(y_train)
y_test = label_encoder.transform(y_test)
```

```python
# Рассмотрим новые классы
label_encoder.classes_
```

Значениям переменной *прежний_уровень* будет присвоен 0, а значениям *снизилась* - 1

```python
# Создаём списки с названиями признаков
ohe_columns = ['популярная_категория', 'маркет_актив_тек_мес']
ord_columns = ['тип_сервиса', 'разрешить_сообщать']
num_columns = ['маркет_актив_6_мес', 'длительность', 'акционные_покупки',  
               'неоплаченные_продукты_штук_квартал', 'ошибка_сервиса', 'средний_просмотр_категорий_за_визит',
               'страниц_за_визит', 'выручка_препредыдущий_месяц', 'минуты_предыдущий_месяц',
                'минуты_текущий_месяц', 'тек_пред_выручка']
```

```python
# Создаём пайплайн для подготовки признаков из списка ohe_columns: заполнение пропусков и OHE-кодирование
# SimpleImputer + OHE
ohe_pipe = Pipeline(
                [
                    ('simpleImputer_ohe', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),
                    ('ohe', OneHotEncoder(drop='first', handle_unknown='ignore', sparse_output=False))
                ]
)

# Создаём пайплайн для подготовки признаков из списка ord_columns: заполнение пропусков и Ordinal-кодирование
# SimpleImputer + OE
ord_pipe = Pipeline(
            [
                ('simpleImputer_before_ord', SimpleImputer(missing_values=np.nan, strategy='most_frequent')),
                ('ord', OrdinalEncoder(categories=[
                                                    ['премиум', 'стандарт'],
                                                    ['да', 'нет']
                                                  ],
                                       handle_unknown='use_encoded_value', unknown_value=np.nan)
                ),
                ('simpleImputer_after_ord', SimpleImputer(missing_values=np.nan, strategy='most_frequent'))
            ]
)
```

```python
# Создаём общий пайплайн для подготовки данных
data_preprocessor = ColumnTransformer(
                            [
                                ('ohe', ohe_pipe, ohe_columns),
                                ('ord', ord_pipe, ord_columns),
                                ('num', MinMaxScaler(), num_columns)
                            ],
                            remainder='passthrough'
)
```

```python
# Создаём итоговый пайплайн: подготовка данных и модель
pipe_final = Pipeline([
                        ('preprocessor', data_preprocessor),
                        ('models', DecisionTreeClassifier(random_state=RANDOM_STATE))
                    ])

param_grid = [

        {
            'models': [KNeighborsClassifier()],
            'models__n_neighbors': range(2,5),
            'models__weights': ['uniform', 'distance'],
            'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']
        },    
        
        {
            'models': [DecisionTreeClassifier(random_state=RANDOM_STATE)],
            'models__max_depth': range(2,5),
            'models__max_features': range(2,5),
            'models__min_samples_split': [2, 5, 10, 20],
            'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']
        },

        {
            'models': [LogisticRegression(random_state=RANDOM_STATE, penalty='l1')],
            'models__C': range(1,5),
            'models__solver': ['liblinear', 'saga'],
            'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']
        },
        
        {
            'models': [SVC(random_state=RANDOM_STATE)],
            'models__C': range(1,5),
            'models__kernel': ['linear', 'rbf', 'poly', 'sigmoid'],
            'preprocessor__num': [StandardScaler(), MinMaxScaler(), 'passthrough']  
        }
    
]
```

```python
# Найдем лучшую модель, используя метрику 'roc_auc'
randomized_search = RandomizedSearchCV(pipe_final, 
                                         param_grid,
                                         cv=5,
                                         scoring='roc_auc',
                                         random_state=RANDOM_STATE,
                                         n_jobs=-1) 

randomized_search.fit(X_train, y_train)
print('Лучшая модель и её параметры:\n', randomized_search.best_estimator_)
```

```python
# Выводим параметры лучшей модели
randomized_search.best_estimator_['preprocessor'].transformers_
```

```python
# Выводим получившиеся метрики модели
print ('Метрика лучшей модели на тренировочной выборке:', round(randomized_search.best_score_, 5))

y_pred_proba = randomized_search.predict_proba(X_test)
print(f'Метрика ROC-AUC на тестовой выборке: {round(roc_auc_score(y_test, y_pred_proba[:, 1]), 5)}')
```

**Вывод по разделу:**
- На этапе предобработки данных использован ColumnTransformer для раздельной обработки количественных и категориальных признаков. 
- Обучены четыре модели (KNeighborsClassifier, DecisionTreeClassifier, LogisticRegression, SVC), для каждой подобрано минимум один гиперпараметр с использованием выбранной метрики оценки. 
- Для выбора лучшей модели применена стратегия использования одного общего пайплайна для всех моделей и инструмента подбора гиперпараметров.
- Проведено исследование влияния различных наборов входных признаков на качество модели с целью её улучшения.
- Лучшая модель и её параметры: LogisticRegression(C=1, penalty='l1', random_state=42, solver='liblinear').
- Метрика лучшей модели на тренировочной выборке: 0.89873.
- Метрика ROC-AUC на тестовой выборке: 0.91854.
